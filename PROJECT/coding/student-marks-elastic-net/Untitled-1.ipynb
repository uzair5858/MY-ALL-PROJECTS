```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = {
    "studytime": [2, 3, 1, 4, 2, 3, 1, 4],
    "failures":  [0, 0, 1, 0, 0, 0, 2, 0],
    "absences":  [4, 2, 8, 1, 5, 3, 10, 0],
    "G1":        [10, 14, 8, 17, 12, 15, 6, 18],
    "G2":        [11, 15, 9, 18, 13, 16, 7, 19],
    "G3":        [11, 15, 8, 19, 13, 16, 6, 20]
}

df = pd.DataFrame(data)
df
```

```python
X = df[["studytime", "failures", "absences", "G1", "G2"]]
y = df["G3"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model = ElasticNet(alpha=1.0, l1_ratio=0.5)
model.fit(X_train, y_train)

print("Elastic Net model trained successfully! ❤️")
```

```python
y_pred = model.predict(X_test)

print("Actual marks:", list(y_test))
print("Predicted marks:", y_pred)
print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
print("Weights:", model.coef_)
```

```python
new_student = pd.DataFrame([{
    "studytime": 4,
    "failures": 0,
    "absences": 1,
    "G1": 17,
    "G2": 18
}])

print("Predicted final marks:", model.predict(new_student)[0])
```
