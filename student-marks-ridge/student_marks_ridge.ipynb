```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, r2_score

```

```python

data = {
    "studytime": [2, 3, 1, 4, 2, 3, 1, 4],
    "failures":  [0, 0, 1, 0, 0, 0, 2, 0],
    "absences":  [4, 2, 8, 1, 5, 3, 10, 0],
    "G1":        [10, 14, 8, 17, 12, 15, 6, 18],
    "G2":        [11, 15, 9, 18, 13, 16, 7, 19],
    "G3":        [11, 15, 8, 19, 13, 16, 6, 20]
}

df = pd.DataFrame(data)
df

```

```python

X = df[["studytime", "failures", "absences", "G1", "G2"]]
y = df["G3"]

print(X)
print(y)

```

```python

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

```

```python

model = Ridge(alpha=1.0)
model.fit(X_train, y_train)

print("Ridge model trained successfully!")

```

```python

y_pred = model.predict(X_test)

print("Actual marks:", list(y_test))
print("Predicted marks:", y_pred)

```

```python

print("MSE:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))

```

```python

new_student = pd.DataFrame([{
    "studytime": 3,
    "failures": 3,
    "absences": 18,
    "G1": 8,
    "G2": 8
}])

print("Predicted final marks:", model.predict(new_student)[0])

```
